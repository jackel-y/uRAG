# ARU Student Support Agent (ARU Genie)

This notebook demonstrates the capabilities of the ARU Genie agentic RAG system. The system can:
1. Answer general questions about ARU academic regulations.
2. Authenticate students using SID and PIN.
3. Provide personalized information (marks, deadlines, etc.) after authentication.

```python
import os
import sys
# Ensure project root is in path
sys.path.append('..')

from src.agent import ARUGenie
from dotenv import load_dotenv

load_dotenv(dotenv_path='../.env')

genie = ARUGenie(vector_store_path='../data/vector_store/faiss_index', student_data_path='../data/student_records/students.csv')
agent_executor = genie.get_agent()
chat_history = []
```

## 1. General Query (No Auth Required)
Asking about the mitigation policy.

```python
response = agent_executor.invoke({"input": "What is the policy for mitigation?", "chat_history": chat_history})
print(response['output'])
```

## 2. Personalized Query (Requires Auth)
Asking for the next deadline.

```python
response = agent_executor.invoke({"input": "When is my next deadline?", "chat_history": chat_history})
print(response['output'])
```

## 3. Authentication
Providing SID: 12345678 and PIN: 4821.

```python
response = agent_executor.invoke({"input": "My SID is 12345678 and my PIN is 4821.", "chat_history": chat_history})
print(response['output'])
# Update history
chat_history.append(("user", "My SID is 12345678 and my PIN is 4821."))
chat_history.append(("assistant", response['output']))
```

## 4. Retrying Personalized Query (Authenticated)
Now the agent should have access to the records.

```python
response = agent_executor.invoke({"input": "Now tell me when is my next deadline?", "chat_history": chat_history})
print(response['output'])
```
